## 2. Introducción teórica

### 2.1 Filtros analógicos y función transferencia

La **función transferencia** es una herramienta matemática que describe el comportamiento de un sistema lineal e invariante en el tiempo (LTI):

$$H(s)=\frac{V_{salida}}{V_{entrada}}=\frac{Y(s)}{X(s)}\qquad \text{con } s=\sigma+j\omega$$

### 2.2 Respuesta en frecuencia

La respuesta en frecuencia describe cómo reacciona el sistema cuando se le dan señales de distintas frecuencias. Se obtiene evaluando la función transferencia sobre el eje imaginario, $s=j\omega$:

$$H(j\omega)=H(s)\big|_{s=j\omega}\;\longrightarrow\;|H(\omega)|,\quad \angle H(\omega)$$

Esto da el módulo $|H(\omega)|$ y la fase $\angle H(\omega)$ del filtro en función de la frecuencia $\omega$.

$$|H|=\sqrt{\text{Re}^2+\text{Im}^2}\qquad\qquad \angle H=\operatorname{arctg}\!\left(\frac{\text{Im}}{\text{Re}}\right)=\sum\angle\text{ceros}-\sum\angle\text{polos}$$



**Estabilidad:** para que el sistema sea estable, todos los polos deben estar en el semiplano izquierdo, es decir, $\text{Re}\{s_p\}<0$.

### 2.3 Polos, ceros y orden

- Los **ceros** son los valores de $s$ donde $H(s)=0$, es decir, donde $V_{salida}=Y(s)=0$ (raíces del numerador).
- Los **polos** son los valores de $s$ donde $H(s)\to\infty$ (raíces del denominador).
- El **orden** está dado por el grado del polinomio del denominador. En circuitos pasivos coincide con la cantidad de elementos reactivos que almacenan energía ($L$ y $C$): si hay $L$ y $C$ es de orden 2; si solo hay uno de ellos, es de orden 1.

### 2.5 Forma canónica de segundo orden y relaciones útiles

Todos los circuitos de esta tarea tienen el mismo denominador:

$$s^2+\frac{\omega_0}{Q}\,s+\omega_0^2$$

con $\omega_0=\dfrac{1}{\sqrt{LC}}$ la frecuencia natural y $Q$ el factor de calidad. Con $Q=\frac{\sqrt2}{2}$ los polos son:

$$s_{1,2}=-\frac{\omega_0}{2Q}\pm j\,\omega_0\sqrt{1-\frac{1}{4Q^2}}=-\frac{\sqrt2}{2}\pm j\frac{\sqrt2}{2}=e^{\pm j\,3\pi/4}$$

Es decir, dos polos complejos conjugados sobre la circunferencia de radio $\omega_0=1$, con un ripple $\zeta=\frac{1}{2Q}=\frac{\sqrt2}{2}$. Para el pasa-bajos esto corresponde a una respuesta **Butterworth** (máximamente plana).

Para los pasa-banda, el ancho de banda a $-3\,\text{dB}$ cumple $BW=\dfrac{\omega_0}{Q}$.

### 3.1 Análisis preliminar

La impedancia serie es $Z_1=sL+R$ y la de la salida es $Z_2=Z_C=\dfrac{1}{sC}$.

- En $\omega=0$: $Z_L=0$ y $Z_C=\infty$ → todo $V_1$ aparece en la salida.
- En $\omega\to\infty$: $Z_L=\infty$ y $Z_C=0$ → la salida se cortocircuita.

Entonces se trata de un **filtro pasa-bajos** (de 2.º orden, porque hay un $L$ y un $C$).

### 3.2 Función de transferencia

$$H(s)=\frac{V_2}{V_1}=\frac{Z_2}{Z_1+Z_2}=\frac{\frac{1}{sC}}{sL+R+\frac{1}{sC}}=\frac{1}{s^2LC+sRC+1}=\frac{\frac{1}{LC}}{s^2+\frac{R}{L}s+\frac{1}{LC}}$$

**Normalización.** Comparando con $\dfrac{\omega_0^2}{s^2+\frac{\omega_0}{Q}s+\omega_0^2}$:

$$\omega_0^2=\frac{1}{LC}=1,\qquad \frac{\omega_0}{Q}=\frac{R}{L}=\sqrt2,\qquad R=1\ \Omega$$

$$\Rightarrow\; L=\frac{RQ}{\omega_0}=\frac{\sqrt2}{2}\ \approx0{,}707\ \qquad C=\frac{1}{\omega_0^2L}=\sqrt2\ \approx1{,}414\ $$

$$\boxed{H(s)=\frac{1}{s^2+\sqrt2\,s+1}}$$

### 3.3 Polos y ceros

- **Ceros:** no hay ceros finitos; ambos están "implícitos" en $s\to\infty$ (grado del denominador − grado del numerador = 2).
- **Polos:**

$$s_1=-\frac{\sqrt2}{2}+j\frac{\sqrt2}{2},\qquad s_2=-\frac{\sqrt2}{2}-j\frac{\sqrt2}{2}$$

Ambos con $\text{Re}<0$ → el sistema es estable.

Con $s=j\omega$:

$$H(j\omega)=\frac{1}{1-\omega^2+j\sqrt2\,\omega}$$

$$\boxed{|H(\omega)|=\frac{1}{\sqrt{(1-\omega^2)^2+2\omega^2}}=\frac{1}{\sqrt{1+\omega^4}}}$$

$$\boxed{\angle H(\omega)=-\operatorname{arctg}\!\left(\frac{\sqrt2\,\omega}{1-\omega^2}\right)}$$

### 6.2 Re-escalado de componentes para $f_0=10\ \text{kHz}$ y $R=50\ \Omega$

**Planteo.** Se busca que la frecuencia natural sea $f_0 = 10\ \text{kHz}$, es decir:

$$\omega_0 = 2\pi f_0 = 2\pi \cdot 10\ \text{kHz} \approx 62\,832\ \text{rad/s}$$

Del denominador de $H(s)$, $\;s^2+\frac{1}{RC}s+\frac{1}{LC}$, y comparándolo con la forma canónica $s^2+\frac{\omega_0}{Q}s+\omega_0^2$, se obtienen dos condiciones:

$$\frac{1}{LC}=\omega_0^2 \qquad\text{y}\qquad \frac{1}{RC}=\frac{\omega_0}{Q}$$

**Cálculo de $C$.** Se elevan al cuadrado ambos miembros de la segunda condición y se usa $Q=\frac{\sqrt2}{2}\Rightarrow\frac{1}{Q^2}=2$:

$$\frac{1}{(RC)^2}=\frac{\omega_0^2}{Q^2}=2\,\omega_0^2=(2\pi\cdot10\ \text{kHz})^2\cdot 2$$

Despejando $C^2$:

$$C^2=\frac{1}{(2\pi\cdot10\ \text{kHz})^2\cdot 2\cdot R^2}\;\Rightarrow\; C=\frac{1}{2\pi\cdot10\ \text{kHz}\cdot\sqrt2\cdot R}$$

$$C=\frac{1}{(2\pi\cdot 10\ \text{kHz})\,\sqrt2\,(50\ \Omega)}\approx 2{,}25\times10^{-7}\ \text{F}=225\ \text{nF}$$

**Cálculo de $L$.** De la primera condición, $\frac{1}{LC}=\omega_0^2$:

$$L=\frac{1}{(2\pi\cdot10\ \text{kHz})^2\cdot C}\approx 1{,}12\times10^{-3}\ \text{H}=1{,}12\ \text{mH}$$

**Resultado:**

$$\boxed{R=50\ \Omega,\qquad C\approx225\ \text{nF},\qquad L\approx1{,}12\ \text{mH}}$$

**En palabras.** Primero se pasa la frecuencia pedida a radianes por segundo. Después se usan las dos relaciones que fijan el comportamiento del circuito: una vincula $L$ y $C$ con la frecuencia natural, y la otra vincula $R$ y $C$ con el factor de calidad $Q$. Como $R=50\ \Omega$ es dato y $Q=\frac{\sqrt2}{2}$, la segunda relación permite despejar $C$. Con ese $C$, la primera relación da $L$.

> Estas relaciones valen para los circuitos 2, 3 y Bonus, donde $\frac{1}{RC}=\frac{\omega_0}{Q}$. En el circuito 1 (pasa-bajos) la condición es $\frac{R}{L}=\frac{\omega_0}{Q}$, por lo que da $L\approx562{,}7\ \mu\text{H}$ y $C\approx450\ \text{nF}$.
