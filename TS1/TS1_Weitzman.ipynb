#### Item 1)

```python
import numpy as np
import matplotlib.pyplot as plt
def mi_funcion_sen(vmax, dc, ff, ph, nn, fs):
    xx=dc+vmax*np.sin(2*np.pi*ff*nn/fs+np.pi*ph)
    tt=nn/fs
    return(tt,xx)


#%% Definiciones para el a)

fs = 20000 # Hz. Queremos que haya 10 muestras en un ciclo en una señal de 2kHz => fs=f_señal*10
# Aclaracion de fs:       fs = f*MPC (muestras por ciclo, en este caso 10)
N = 1000 # muestras (me las dan)
vmax=np.sqrt(2) # pot = 1
dc=0 # Valor medio (alrededor del que oscila)

# buscamos la resolucion espectral --> df=20'000/1000 = 20
df=fs/N # Resolucion Espectral

# buscamos el k: f=k*df --> 2000=k*20 --> k=100
k = 100 # frecuencia de bin
kk = np.arange(0,k)
kdf=k*df # es nuestra f deseada = 2000Hz

ph=0
nn=np.arange(0,N)
mu=0 #valor medio
tt, xx= mi_funcion_sen(vmax, dc, kdf, ph, nn, fs)
modulo_trx=np.abs(np.fft.fft(xx)/N)
dB_modulo_trx = 20*np.log10(modulo_trx)

#%% Grafico 1

plt.figure()
plt.plot(tt[:50],xx[:50], ':o', color="blue")
plt.title("Grafico 1", fontdict=None, loc=None, pad=None)
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud [V]")
plt.grid(True) #Para mostrar una grilla de fondo
plt.show()

plt.figure()
plt.title("Grafico 2", fontdict=None, loc=None, pad=None)
plt.plot(dB_modulo_trx[:N//2+1], color="blue")
plt.xlabel("Frecuencia en bins [k]")
plt.ylabel("Amplitud [dB]")
plt.grid(True) #Para mostrar una grilla de fondo
plt.show()


```

Si ploteamos algunos ciclos de una sinusoidal de 2KHz con 10 muestras por ciclo y luego ploteamos el espectro de frecuencia hasta la frecuencia de Nyquist, obtenemos los gráficos 1 y 2. Notamos que en k = 100 está concentrada la energía de la onda total. El ruido restante se debe a el redondeo por el numero de mantisa que es, por supuesto, finito, y no con la señal en sí.  
También es interesante notar que normalizamos nuestra transformada de fourier con un factor de 1/N, lo que permite que no se amplifique los valores del módulo de cada frecuencia en el espectro.

Acá lo unico que buscamos fue cambiar la potencia de la señal y desfasarla. La potencia en una sinusoidal se calcula como


$P_{sin} = \frac{V_{max}^2}{2}$

que es la varianza. Ambas coinciden si $μ = 0$ (la media)

Vemos también que el ruido no está correlacionado con la fase ni con la potencia de la señal como se puede apreciar en el gráfico 4.2. Notamos que al hacer la diferencia de los espectros de ambas senoidales, lo único que sobrevive son los 0,3V de potencia de diferencia que había entre ambas señales, pero el ruido podemos confirmar que no se ve afectado por el cambio de fase y de potencia de la señal.

```python
##% item c)
import scipy.stats as stats
# Una secuencia aleatoria de ruido normalmente distribuido con DC (valor medio) 0V y varianza 0.1 W.
def ruidoNormal(N,mu,pot_ruido):
  yy = np.random.normal(mu,pot_ruido,N)
  return (yy)
mu = 0 # [V]
pot_ruido = 0.1 # [W]
sigma = np.sqrt(pot_ruido)
yy = ruidoNormal(N, mu, sigma)
modulo_tr = np.abs(np.fft.fft(yy)/N)


#

#%% Grafico 5
plt.figure()
plt.title("Grafico 5", fontdict=None, loc=None, pad=None)

plt.plot(tt[:50],yy[:50], ':o', color="blue")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud [V]")
plt.grid(True) #Para mostrar una grilla de fondo
plt.show()

#%% Grafico 6
plt.figure()
plt.title("Grafico 6", fontdict=None, loc=None, pad=None)

plt.plot(modulo_tr[:N//2+1], color="blue")
plt.xlabel("Frecuencia en bins [k]")
plt.ylabel("Amplitud [V]")
plt.grid(True) #Para mostrar una grilla de fondo
plt.show()


# Grafico 7 (Histograma y Gaussiana, ayuda con IA Gemini)

plt.figure(figsize=(8, 5))
# Histograma normalizado (Densidad de probabilidad empírica)
plt.hist(yy, bins=30, density=True, alpha=0.6, color="blue", label="Empírica (Ruido generado)")

# Curva teórica normal en rojo
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 200)
p = stats.norm.pdf(x, mu, sigma)

plt.title("Distribución de Amplitudes vs. PDF Teórica (Ruido)")
#
plt.plot(x, p, color="red", linewidth=2.5, label="Teórica (PDF Gaussiana)")
plt.xlabel("Amplitud [V]")
plt.ylabel("Densidad de Probabilidad")
plt.legend()
plt.grid(True)
plt.show()
```

En este inciso se buscó graficar una onda de ruido con una distribución Normal de amplitud 0,1V y una media de 0V.

Podemos apreciar en el histograma que sigue la tendencia de la línea teórica roja con bastante fidelidad, algo que seguro puede ajustarse más si se aumenta la cantidad de Bins que plotea el gráfico (* Corrección más adelante. No quise borrar esto porque es evidencia de aprendizaje)

```python
##% item d)
# Una secuencia aleatoria de ruido normalmente distribuido con DC (valor medio) 0V y varianza 0.1 W.
def ruidoUniforme(N,mu,pot_ruido):
  # Varianza_uniforme = (b-a)^2/12 = b^2/3 = 0.1
  # sqrt(0.1 * 3) = b
  limite = np.sqrt(3*pot_ruido)
  print(limite)
  yy = np.random.uniform(-limite,limite,N)
  return (yy)
mu = 0 # [V]
pot_ruido = 0.1 # [W]
yy = ruidoUniforme(N, mu, pot_ruido)


# Límites teóricos para el gráfico analítico
limite_teorico = np.sqrt(3 * pot_ruido)
pdf_teorica = 1 / (2 * limite_teorico)  # Altura de la PDF uniforme: 1 / (b - a)

#%% Grafico 8
plt.figure()
plt.title("Grafico 8", fontdict=None, loc=None, pad=None)
plt.plot(tt[:100],yy[:100], ':o', color="blue")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud [V]")
plt.grid(True) #Para mostrar una grilla de fondo
plt.show()

modulo_tr=np.abs(np.fft.fft(yy)/N)

# Grafico 9
plt.figure()
plt.title("Grafico 9", fontdict=None, loc=None, pad=None)
plt.plot(modulo_tr[:N//2], color="blue")
plt.xlabel("Frecuencia en bins [k]")
plt.ylabel("Amplitud [V]")
plt.grid(True) #Para mostrar una grilla de fondo
plt.show()

# A partir de acá me ayudé con la IA
plt.figure(figsize=(8, 5))
plt.title("Grafico 10: D. de Amplitudes vs. Teórica Uniforme")

# Histograma normalizado (Densidad de probabilidad empírica)
plt.hist(yy, bins=10, density=True, alpha=0.6, color="blue", label="Empírica (Ruido generado)")

# Curva teórica uniforme en rojo (Forma de pulso rectangular)
x_teorico = [-limite_teorico, -limite_teorico, limite_teorico, limite_teorico]
y_teorico = [0, pdf_teorica, pdf_teorica, 0]
plt.plot(x_teorico, y_teorico, color="red", linewidth=2.5, label="Teórica (PDF Uniforme)")

plt.yticks(np.arange(0, 1.1, 0.1))
plt.grid(True, which='major', axis='y', linestyle='--', color='gray', alpha=0.3)

plt.xlim(-limite_teorico * 1.3, limite_teorico * 1.3)
plt.xlabel("Amplitud [V]")
plt.ylabel("Densidad de Probabilidad")
plt.legend()
plt.grid(True)
plt.show()
```

Este gráfico me resultó de particularidad interés, al igual que el gráfico anterior, y es que aprendí un poco más de las distribuciones de probabilidad. Sabía que la densidad de probabilidad debía integrar a 1, entonces me hacía ruido que las densidades pudieran tener una amplitud de poco más o un poco menos que 1. Me hacía "ruido". Después de pensarlo un rato, se puede apreciar que la integral, geométricamente, es el área bajo la curva. Entonces, en una distribución normal,

$Área = Base.Altura = 1$

Por ende,

$Altura = \frac{1}{2*limite} ≈ \frac{1}{2*0.548} ≈ 0,912$

que es perfectamente lo que observamos en el gráfico.

En la distribución anterior, la altura máxima debía dar:

$A = \frac{1}{σ*\sqrt{2π}}$, y como $σ = \sqrt{0,1} ≈ 0,3162$, entonces:

$A ≈ 1,26$

que es exactamente lo que da el valor de la linea teórica en su máximo.

##**En cuanto a los Bins**

Me resultó de sumo interés la siguiente observación. Creía (en la distribución normal) que al aumentar la cantidad de Bins, o casilleros posibles para contar la cantidad de veces que una amplitud de ruido caía en cierta frecuencia, la distribución de densidad se acercaría más a la teórica. En otras palabras, que tendría menos error absoluto cada columna con respecto a su curva teórica. Al intentarlo, notaba que el error aumentaba considerablemente. Esto se debe a que la probabilidad que columnas menos anchas reciban una cantidad similar de muestras dismiuye considerablemente. Por el contrario, en una distribución Normal, una baja cantidad de Bins representada en el Histograma, no representaría de manera notable la tendencia Gaussiana del comportamiento del ruido. En una distribución uniforme, este Histograma debería comportarse mejor con menor cantidad de Bins, aunque no sé hasta qué valor de Bins continua esa tendencia. Sería un interesante trabajo analizar la curva de la suma de los errores con respecto al valor teórico en funcion de los Bins y encontrar el valor de Bins para que esa curva alcance un mínimo.
```python
# Un pulso rectangular de la misma frecuencia, 1 W de potencia y ciclo de actividad del 50%
from scipy import signal

cuadrada = signal.square(2*np.pi*kdf*nn/fs,0.5)

#%% Grafico 11
plt.figure()
plt.title("Grafico 11", fontdict=None, loc=None, pad=None)
plt.plot(tt[:50],cuadrada[:50], ':o', color="blue")
plt.xlabel("Tiempo [s]")
plt.ylabel("Amplitud [V]")
plt.grid(True) #Para mostrar una grilla de fondo
plt.show()

modulo_tr=np.abs(np.fft.fft(cuadrada)/N)
dB_modulo_tr = 20*np.log10(modulo_tr)

#%% Grafico 12
plt.figure()
plt.title("Grafico 12", fontdict=None, loc=None, pad=None)
plt.plot(modulo_tr[:N//2+1], color="blue")
plt.xlabel("Frecuencia en bins [k]")
plt.ylabel("Amplitud [V]")
plt.grid(True) #Para mostrar una grilla de fondo
plt.show()


#%% Grafico 13
plt.figure()
plt.title("Grafico 13", fontdict=None, loc=None, pad=None)
plt.plot(dB_modulo_tr[:N//2+1], color="blue")
plt.xlabel("Frecuencia en bins [k]")
plt.ylabel("Amplitud [dB]")
plt.grid(True) #Para mostrar una grilla de fondo
plt.show()
```

Acá podemos apreciar la naturaleza de 2 fenómenos: la transformada de Fourier de una onda cuadrada y el derrame frecuencial.

Sabemos que la TF de una onda cuadrada son los multiplos **impares** de la frecuencia fundamental, si no me equivoco, con una caida exponencial en cada una de sus amplitudes. La DFT genera que al tomar muestras en el tiempo con una forma cuadrada, en cada armónico estamos generando una Sinc. Esto se podría cambiar con un ventaneo, algo que vamos a ver más adelante pero estuve investigando un poco (muy interesante me resultó).

El derrame frecuencial, en este caso, es esa Sinc antes mencionada. Los picos caen en 100, 300, 500 bins ya que mi $Δf = 20Hz$ entonces en el Bin 100 "cae" mi frecuencia de onda cuadrada fundamental, unos 2KHz. El resto son esos múltiplos impares de la TF. Se puede observar en el gráfico de amplitud en dB (Gráfico 13) cómo la Sinc genera un derrame alrededor de las frecuencias de la TF, y cómo en este gráfico logarítmico las amplitudes de las frecuencias principales caen de manera similar a una lineal, lo que se corresponde con lo anticipado previamente.
